```python
# ==========================================
# 1. IMPORT LIBRARIES
# ==========================================

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import time
```

```python
# ==========================================
# 2. LOAD DATASET
# ==========================================

df = pd.read_csv("used_phone_price_prediction_1M.csv")

print("Dataset shape:", df.shape)
print(df.head())
```

```python
# ==========================================
# 3. REMOVE DUPLICATES
# ==========================================

print("Duplicates before:", df.duplicated().sum())

df.drop_duplicates(inplace=True)

print("Duplicates after:", df.duplicated().sum())
```

```python
# ==========================================
# 4. HANDLE MISSING VALUES
# ==========================================

# Numerical columns
numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())


# Categorical columns
categorical_columns = df.select_dtypes(
    include=['object', 'category']
).columns

for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])


print("Total missing values:", df.isnull().sum().sum())
```

```python
# ==========================================
# 5. FEATURE ENGINEERING
# ==========================================



# Remove unwanted columns
df.drop(
    columns=[
        'id',
        'age_months',
        'calculated_age_months',
    ],
    errors='ignore',
    inplace=True
)


print("Final dataset shape:", df.shape)
print(df.columns.tolist())
```

```python
# ==========================================
# 6. INPUT FEATURES AND TARGET
# ==========================================

X = df.drop(columns=['resale_price'])

y = df['resale_price']


print("Input shape:", X.shape)
print("Target shape:", y.shape)
```

```python
# ==========================================
# 7. REDUCE MEMORY USAGE
# ==========================================

num_cols_temp = X.select_dtypes(include=np.number).columns

for col in num_cols_temp:
    X[col] = X[col].astype(np.float32)


print(X.dtypes)
```

```python
# ==========================================
# 8. IDENTIFY COLUMN TYPES
# ==========================================

categorical_cols = X.select_dtypes(
    include=['object', 'category']
).columns.tolist()

numerical_cols = X.select_dtypes(
    include=np.number
).columns.tolist()


print("Categorical columns:")
print(categorical_cols)

print("\nNumerical columns:")
print(numerical_cols)
```

```python
print("Unique brands:", df['brand'].nunique())
print("Unique models:", df['model'].nunique())
```

```python
# ==========================================
# 9. TRAIN TEST SPLIT
# ==========================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)


print("Training shape:", X_train.shape)
print("Testing shape :", X_test.shape)
```

```python
# ==========================================
# 10. LINEAR REGRESSION PREPROCESSOR
# ==========================================

lr_preprocessor = ColumnTransformer(
    transformers=[
        (
            'num',
            StandardScaler(),
            numerical_cols
        ),

        (
            'cat',
            OneHotEncoder(
                handle_unknown='ignore',
                dtype=np.float32
            ),
            categorical_cols
        )
    ]
)
```

```python
# ==========================================
# 11. TRAIN LINEAR REGRESSION
# ==========================================

lr_model = Pipeline([
    ('preprocessor', lr_preprocessor),
    ('model', LinearRegression())
], verbose=True)


start = time.time()

lr_model.fit(X_train, y_train)

print(
    "Linear Regression training time:",
    round(time.time() - start, 2),
    "seconds"
)


y_pred_lr = lr_model.predict(X_test)
```

```python
# ==========================================
# 12. TREE PREPROCESSOR
# ==========================================

tree_preprocessor = ColumnTransformer(
    transformers=[
        (
            'cat',
            OneHotEncoder(
                handle_unknown='ignore',
                dtype=np.float32
            ),
            categorical_cols
        )
    ],
    remainder='passthrough'
)
```

```python
# ==========================================
# 13. TRAIN DECISION TREE
# ==========================================

dt_model = Pipeline([
    ('preprocessor', tree_preprocessor),

    (
        'model',
        DecisionTreeRegressor(
            max_depth=20,
            min_samples_split=5,
            min_samples_leaf=2,
            random_state=42
        )
    )
], verbose=True)


start = time.time()

dt_model.fit(X_train, y_train)

print(
    "Decision Tree training time:",
    round(time.time() - start, 2),
    "seconds"
)


y_pred_dt = dt_model.predict(X_test)
```

```python
# ==========================================
# 14. RANDOM FOREST
# ==========================================

rf_model = Pipeline([
    ('preprocessor', tree_preprocessor),

    (
        'model',
        RandomForestRegressor(
            n_estimators=50,
            max_depth=15,
            min_samples_split=10,
            min_samples_leaf=4,
            max_features=0.7,
            random_state=42,
            n_jobs=2,
            verbose=2
        )
    )
], verbose=True)


start = time.time()

rf_model.fit(X_train, y_train)

print(
    "Random Forest training time:",
    round(time.time() - start, 2),
    "seconds"
)


y_pred_rf = rf_model.predict(X_test)
```

```python
# ==========================================
# 15. EVALUATION FUNCTION
# ==========================================

def evaluate_model(name, y_true, y_pred):

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    mse = mean_squared_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        y_true,
        y_pred
    )

    return {
        'Model': name,
        'MAE': mae,
        'MSE': mse,
        'RMSE': rmse,
        'R2 Score': r2
    }
```

```python
# ==========================================
# 16. COMPARE ALL MODELS
# ==========================================

results = [

    evaluate_model(
        "Linear Regression",
        y_test,
        y_pred_lr
    ),

    evaluate_model(
        "Decision Tree",
        y_test,
        y_pred_dt
    ),

    evaluate_model(
        "Random Forest",
        y_test,
        y_pred_rf
    )
]


results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by='R2 Score',
    ascending=False
)

print(results_df)
```

```python
new_phone = pd.DataFrame([{
    'brand': 'Samsung',
    'model': 'Galaxy S23',
    'release_year': 2023,
    'ram_gb': 8,
    'storage_gb': 256,
    'screen_size_inches': 6.1,
    'battery_capacity': 3900,
    'processor_score': 88,
    'camera_score': 85,
    'os_type': 'Android',
    'has_5g': 1,
    'original_price': 75000,
    'purchase_year': 2024,
    'usage_hours_per_day': 5.0,
    'condition': 'Good',
    'battery_health': 100,
    'screen_cracked': 0,
    'body_damage': 0,
    'repair_history': 0,
    'water_damage': 0,
    'city_tier': 'Tier1',
    'seller_type': 'Individual',
    'warranty_remaining_months': 0,
    'box_available': 1,
    'charger_available': 1,
    'market_demand_score': 8
}])

predicted_price = rf_model.predict(new_phone)

print("Predicted Resale Price:", round(predicted_price[0], 2))
```
